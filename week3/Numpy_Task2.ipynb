```python
A = numpy.random.randint(0, 10, (3, 3))
B = numpy.random.randint(0, 10, (3, 3))
print("Array A:")
print(A)
print("Array B:")
print(B)
equal = numpy.array_equal(A, B)
print("Are they equal?", equal)
```

```python
coordinates = numpy.random.random((100, 2))
distances = numpy.zeros((100, 100))
for i in range(100):
    for j in range(100):
        x1 = coordinates[i][0]
        y1 = coordinates[i][1]
        x2 = coordinates[j][0]
        y2 = coordinates[j][1]
        distances[i][j] = numpy.sqrt((x1 - x2)**2 + (y1 - y2)**2)
print("Coordinates:")
print(coordinates)
print("Distances:")
print(distances)
```

```python
matrix = numpy.random.random((5, 5))
print("Original:")
print(matrix)
row_means = matrix.mean(axis=1)
result = matrix - row_means[:, None]
print("Matrix after subtracting row means:")
print(result)
```

```python
array = numpy.array([
    [5, 2, 8],
    [1, 7, 4],
    [3, 1, 6],
    [2, 9, 5]
])
print("Original:")
print(array)
n = 1
sort = array[array[:, n].argsort()]
print("Sorted array:")
print(sort)
```

```python
matrix = numpy.random.randint(0, 10, (4, 4))
print("Matrix:")
print(matrix)
rank = numpy.linalg.matrix_rank(matrix)
print("Rank:", rank)
```

```python
array = numpy.random.randint(0, 10, (16, 16))
print("Original:")
print(array)
blocks = array.reshape(4, 4, 4, 4)
block_sum = blocks.sum(axis=(1, 3))
print("block sum:")
print(block_sum)
```
