```python
# Numpy Task Part 1  Ashley Eshghi
import numpy
vector = numpy.arange(10, 50)
print("Original vector:")
print(vector)
reverse_vector = vector[::-1]
print("Reversed vector:")
print(reverse_vector)
```

```python
array = numpy.random.random((5, 5))
print("Array:")
print(array)
print("Minimum:", array.min())
print("Maximum:", array.max())
```

```python
matrix = numpy.random.random((5, 5))
print("Original matrix:")
print(matrix)
normalized = (matrix - matrix.min()) / (matrix.max() - matrix.min())
print("Normalize:")
print(normalized)
```

```python
first = numpy.random.random((5, 3))
second = numpy.random.random((3, 2))
print("Frist Matrix:")
print(first)
print("Matrix B:")
print(second)
result = numpy.dot(first,second)
print("Second Matrix:")
print(result)
```

```python
today = numpy.datetime64('today','D')
yesterday = today - numpy.timedelta64(1,'D')
tomorrow = today + numpy.timedelta64(1,'D')
print("Yesterday:", yesterday)
print("Today:", today)
print("Tomorrow:", tomorrow)
```

```python
array = numpy.random.random(5) * 10
print("Original:")
print(array)
print("Method 1:", array - array % 1)
print("Method 2:", array // 1)
print("Method 3:", numpy.floor(array))
print("Method 4:", array.astype(int))
print("Method 5:", numpy.trunc(array))
```
